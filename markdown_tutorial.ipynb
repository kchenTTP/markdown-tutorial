# **Markdown Tutorial**

**Markdown** is a lightweight markup language for adding formatting to plain text documents. With simple characters, you can create headers, lists, links, and more. It is easy to read and write, and supports inline HTML for advanced formatting.

A Markdown parser converts these symbols into HTML, which is then rendered visually by a browser or application. In Google Colab, text cells have a built-in renderer that displays the formatted output when you exit the cell.

```markdown
* List item
  * Nested list item
    * Nested list item
* List item

- List item
  - Nested list item
    - Nested list item
- List item

+ List item
  + Nested list item
    + Nested list item
+ List item
```

Enable language-specific syntax highlighting by adding the language name after the opening ` ``` `:

#### **Python**

```python
def count_to_x(number: int) -> None:
  for n in range(1, number + 1):
    print(n)

count_to_x(10)
```

#### **JavaScript**

```javascript
function countToX(number) {
  for (let n = 1; n <= number; n++) {
    console.log(n);
  }
}

countToX(10);
```

### **What is not allowed?**

- `<script>` tags will not be executed.
- External CSS stylesheets are not supported.
- Interactive elements like HTML forms and buttons won't function within a Markdown rendering context.

## **Misc.**

### **Escape Characters**

Since Markdown uses symbols for formatting, prefix any literal symbol with a backslash (`\`) to prevent it from being interpreted as formatting.

<br>

| Character | Markdown | Preview |
| :-------- | :------- | :-----: |
| `\` | `\\text` | \\text |
| `` ` `` | `` \`text `` | \`text |
| `*` | `\*text` | \*text |
| `_` | `\_text` | \_text |
| `{}` | `\{text\}` | \{text\} |
| `[]` | `\[text\]` | \[text\] |
| `()` | `\(text\)` | \(text\) |
| `#` | `\#text` | \#text |
| `+` | `\+text` | \+text |
| `-` | `\-text` | \-text |
| `.` | `\.text` | \.text |
| `!` | `\!text` | \!text |
| `$` | `\$text` | \$text |


### **Manual Line Breaks**

To force a line break, add two trailing spaces to the end of a line.

```markdown
No trailing spaces.
Everything is on the same line.

With trailing spaces.<space><space>
On a different line.
```

No trailing spaces.
Everything is on the same line.

With trailing spaces.  
On a different line.
